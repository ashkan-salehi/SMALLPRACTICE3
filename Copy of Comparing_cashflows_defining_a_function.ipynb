```python

# define a function to compare cashflow 1 and cashflow 2
def compare(cf1, cf2, rate):
  NPV1=0
  NPV2=0
  for cashflow in range(len(cf1)):
    NPV1 += cf1[cashflow]/(1+rate)**cashflow
  for cashflow in range(len(cf2)):
    NPV2 += cf2[cashflow]/(1+rate)**cashflow
  if NPV1 > NPV2:
    print(f"for rate {rate} , the NPV1 is {NPV1: .2f}.")
  if NPV2 > NPV1:
    print(f"for rate {rate} , the NPV2 is {NPV2: .2f}.")


# insert cashflows and the rate
cf1=[float(x) for x in input("enter your list: ").split()]
cf2=[float(x) for x in input("enter your list: ").split()]
ratestring= input("please enter your rate: ")
rate=float(ratestring)

# use the defined function to run the comparison
compare(cf1, cf2, rate)


```
